# Freqtrade Strategy Analysis
Use this notebook to visualize your backtest results found in `user_data/backtest_results`.

```python
import json
import pandas as pd
import plotly.graph_objects as go
from pathlib import Path

# Load latest backtest result
results_dir = Path('../backtest_results')
files = sorted(results_dir.glob('backtest-result-*.json'))
if files:
    latest_file = files[-1]
    print(f"Loading {latest_file}...")
    with open(latest_file, 'r') as f:
        data = json.load(f)
    
    # Extract strategy stats
    strat_name = list(data['strategy'].keys())[0]
    metrics = data['strategy'][strat_name]
    print(f"Strategy: {strat_name}")
    print(f"Total Profit: {metrics['profit_total_abs']}")
else:
    print("No backtest results found. Run 'freqtrade backtesting' first.")
```
